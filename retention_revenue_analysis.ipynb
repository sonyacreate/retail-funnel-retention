# Проект 2: Customer Retention & Revenue Analytics

## Вопрос исследования
Почему клиенты не возвращаются за повторными покупками и какие сегменты клиентов формируют основную выручку?

## Бизнес-контекст
Компания хочет понять поведение клиентов после первой покупки: как часто они возвращаются, как меняется удержание со временем и какие группы клиентов наиболее ценны для бизнеса.

## Гипотезы
- H1: Значительная часть клиентов совершает только одну покупку и не возвращается в течение периода наблюдения.
- H2: Retention снижается по мере увеличения времени после первой покупки.
- H3: Клиенты с более высокой частотой и суммой покупок формируют непропорционально большую долю выручки.

## Повторные покупки

Здесь мы не называем результат «воронкой»: это распределение клиентов по частоте покупок.

Ключевая метрика — **repeat purchase rate**: доля клиентов, совершивших как минимум две отдельные покупки.

Если доля клиентов с одной покупкой высока, это сигнал для дальнейшего анализа retention: важно понять, в какой период после первой покупки клиенты перестают возвращаться.

```python
# Первая подтверждённая покупка каждого клиента определяет его когорту.
first_purchase = sales.groupby('Customer ID')['InvoiceDate'].min().rename('FirstPurchaseDate')
first_purchase = first_purchase.dt.to_period('M').rename('CohortMonth').reset_index()
first_purchase.head()
```

```python
cohort_sales = sales.merge(first_purchase, on='Customer ID', how='left')
cohort_sales.head()
```

```python
cohort_sales['OrderMonth'] = cohort_sales['InvoiceDate'].dt.to_period('M')
cohort_sales['CohortIndex'] = (
    cohort_sales['OrderMonth'].dt.year * 12 + cohort_sales['OrderMonth'].dt.month
    - cohort_sales['CohortMonth'].dt.year * 12 - cohort_sales['CohortMonth'].dt.month
)
cohort_sales[['Customer ID', 'CohortMonth', 'OrderMonth', 'CohortIndex']].head()
```

```python
cohort_data = (
    cohort_sales.groupby(['CohortMonth', 'CohortIndex'])['Customer ID']
    .nunique()
    .reset_index()
)
cohort_data.head()
```

```python
cohort_pivot = cohort_data.pivot_table(
    index='CohortMonth', columns='CohortIndex', values='Customer ID'
)
cohort_pivot
```

```python
cohort_size = cohort_pivot[0]
retention_matrix = cohort_pivot.divide(cohort_size, axis=0) * 100
retention_matrix.round(1)
```

```python
# Последние когорты имеют меньше месяцев наблюдения, поэтому сравнивать их с ранними
# когортами на длинном горизонте некорректно.
max_observed_month = cohort_sales['OrderMonth'].max()
cohort_age = (
    (max_observed_month.year - retention_matrix.index.year) * 12
    + (max_observed_month.month - retention_matrix.index.month)
)

mature_cohorts = retention_matrix.index[cohort_age >= 5]
retention_mature = retention_matrix.loc[mature_cohorts]
retention_mature.round(1)
```

```python
plt.figure(figsize=(18, 8))
sns.heatmap(
    retention_mature,
    annot=True,
    fmt='.1f',
    cmap='Blues',
    vmin=0,
    vmax=100
)
plt.title('Monthly Customer Retention by Cohort')
plt.xlabel('Months since first purchase')
plt.ylabel('Cohort month')
plt.show()
```

## Retention-матрица: как интерпретировать

Retention в месяце 0 равен 100% по определению: это размер когорты в момент первой покупки.

Для месяца 1 смотрим, какая доля клиентов из конкретной когорты совершила покупку в следующем календарном месяце. Аналогично интерпретируются последующие месяцы.

Важно: последние когорты имеют меньше времени для наблюдения, поэтому их нельзя напрямую сравнивать с ранними когортами на длинном горизонте. Для долгосрочного сравнения выше оставлены только когорты с достаточной зрелостью.

На этом этапе мы описываем поведение клиентов, но не утверждаем причинность. Причины низкого retention потребуют дополнительного анализа.

## SQL-анализ на уровне клиента

После когортного анализа переходим от поведения когорт к уровню отдельного клиента.

Здесь SQL нужен не ради демонстрации синтаксиса: он собирает customer-level витрину, на которой можно принимать решения о repeat-поведении, реактивации и концентрации выручки.

Цель блока — ответить на три бизнес-вопроса:
1. Какова доля one-time и repeat клиентов и какую долю выручки они формируют?
2. Какие клиенты давно не покупали, но исторически приносили много выручки?
3. Насколько концентрирована выручка среди наиболее ценных клиентов?

На этом этапе одна строка в `orders` = один заказ, а одна строка в `customer_metrics` = один клиент.

```python
orders = duckdb.sql("""
SELECT
    "Customer ID",
    Invoice,
    MIN(InvoiceDate) AS order_date,
    SUM(revenue) AS order_revenue
FROM sales
GROUP BY
    "Customer ID",
    Invoice
""").df()

orders.head()
```

```python
customer_metrics = duckdb.sql("""
SELECT
    "Customer ID",
    COUNT(Invoice) AS orders_count,
    SUM(order_revenue) AS total_revenue,
    AVG(order_revenue) AS avg_order_value,
    MIN(order_date) AS first_purchase,
    MAX(order_date) AS last_purchase
FROM orders
GROUP BY "Customer ID"
""").df()

customer_metrics.head()
```

```python
repeat_revenue = duckdb.sql("""
SELECT
    CASE
        WHEN orders_count = 1 THEN 'One-time'
        WHEN orders_count >= 2 THEN 'Repeat'
    END AS customer_type,
    COUNT(*) AS customers,
    SUM(total_revenue) AS total_revenue,
    SUM(total_revenue) * 100.0
        / SUM(SUM(total_revenue)) OVER () AS revenue_share_pct
FROM customer_metrics
GROUP BY customer_type
ORDER BY customers DESC
""").df()

repeat_revenue
```

### Что показывает этот запрос

COUNT(*) после GROUP BY считает клиентов внутри каждой группы, а SUM(SUM(total_revenue)) OVER () получает общую выручку по всем группам.

Таким образом, revenue_share_pct показывает не долю клиентов, а долю общей выручки, которую формирует каждый тип клиента.

```python
customer_segments = duckdb.sql("""
WITH data_end AS (
    SELECT MAX(order_date) AS data_end_date
),
customer_base AS (
    SELECT
        cm."Customer ID",
        cm.orders_count,
        cm.total_revenue,
        cm.avg_order_value,
        cm.first_purchase,
        cm.last_purchase,
        DATE_DIFF('day', cm.last_purchase, de.data_end_date) AS recency_days
    FROM customer_metrics cm
    CROSS JOIN data_end de
),
customer_thresholds AS (
    SELECT
        *,
        MEDIAN(total_revenue) OVER () AS median_revenue
    FROM customer_base
)
SELECT
    "Customer ID",
    orders_count,
    total_revenue,
    avg_order_value,
    first_purchase,
    last_purchase,
    recency_days,
    median_revenue,
    CASE
        WHEN total_revenue >= median_revenue
             AND recency_days >= 90
            THEN 'High-value inactive'
        WHEN recency_days >= 90
            THEN 'Low-value inactive'
        ELSE 'Active'
    END AS customer_segment
FROM customer_thresholds
ORDER BY total_revenue DESC
""").df()

customer_segments.head()
```

### Клиенты с высокой ценностью и длительным отсутствием покупок

Recency считается относительно последней даты в датасете, а не относительно сегодняшнего дня. Это важно: мы анализируем исторический период наблюдения и не смешиваем его с текущей датой.

Порог 90 дней — аналитическое правило для сегментации, а не доказанный порог оттока. Его можно изменить при наличии бизнес-контекста или исторических данных о цикле повторной покупки.

```python
segment_summary = duckdb.sql("""
SELECT
    customer_segment,
    COUNT(*) AS customers,
    SUM(total_revenue) AS total_revenue,
    AVG(total_revenue) AS avg_customer_revenue,
    AVG(orders_count) AS avg_orders
FROM customer_segments
GROUP BY customer_segment
ORDER BY total_revenue DESC
""").df()

segment_summary
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=customer_segments,
    x='recency_days',
    y='total_revenue',
    hue='customer_segment',
    alpha=0.6
)

plt.axvline(90, linestyle='--', label='90 days')
plt.axhline(
    customer_segments['median_revenue'].iloc[0],
    linestyle='--',
    label='Median revenue'
)

plt.title('Customer Recency vs Revenue')
plt.xlabel('Days since last purchase')
plt.ylabel('Total Revenue')
plt.legend()
plt.show()
```

```python
customer_revenue_concentration = duckdb.sql("""
WITH ranked_customers AS (
    SELECT
        "Customer ID",
        total_revenue,
        SUM(total_revenue) OVER (
            ORDER BY total_revenue DESC
            ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
        ) AS cumulative_revenue
    FROM customer_metrics
),
total_revenue AS (
    SELECT SUM(total_revenue) AS total_revenue
    FROM customer_metrics
)
SELECT
    rc."Customer ID",
    rc.total_revenue,
    rc.cumulative_revenue,
    rc.cumulative_revenue * 100.0 / tr.total_revenue
        AS cumulative_revenue_share_pct
FROM ranked_customers rc
CROSS JOIN total_revenue tr
ORDER BY rc.total_revenue DESC
""").df()

customer_revenue_concentration['customer_share_pct'] = (
    (customer_revenue_concentration.index + 1)
    / len(customer_revenue_concentration)
    * 100
)

customer_revenue_concentration.head()
```

```python
# Доля top 10% и top 20% считаем по фактическому числу клиентов.
n_customers = len(customer_revenue_concentration)
top_10_n = max(1, int(np.ceil(n_customers * 0.10)))
top_20_n = max(1, int(np.ceil(n_customers * 0.20)))

top_10_revenue = customer_revenue_concentration.iloc[top_10_n - 1]['cumulative_revenue_share_pct']
top_20_revenue = customer_revenue_concentration.iloc[top_20_n - 1]['cumulative_revenue_share_pct']

print(f"Top 10% of customers account for {top_10_revenue:.1f}% of revenue")
print(f"Top 20% of customers account for {top_20_revenue:.1f}% of revenue")
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    customer_revenue_concentration['customer_share_pct'],
    customer_revenue_concentration['cumulative_revenue_share_pct']
)

plt.axhline(80, linestyle='--', label='80% of revenue')
plt.axvline(20, linestyle='--', label='20% of customers')

plt.title('Customer Revenue Concentration')
plt.xlabel('Share of Customers (%)')
plt.ylabel('Cumulative Share of Revenue (%)')
plt.legend()
plt.show()
```

# Итоговые бизнес-выводы

## Что показывает анализ

### Удержание и повторные покупки
В выборке 5 878 идентифицированных клиентов. 4 255 клиентов (72,4%) совершили как минимум две отдельные покупки, а 1 623 клиента (27,6%) остались one-time. При этом repeat-клиенты сформировали 96,8% общей выручки против 3,2% у one-time клиентов.

### Cohort retention
У зрелых когорт медианный retention в месяце 1 составляет 19,9%, в месяце 5 — 17,9%, в месяце 10 — 11,5%. При этом динамика не является монотонной: в первые месяцы значения колеблются примерно около 20%, поэтому H2 корректнее трактовать как наличие снижения на более длинном горизонте, а не как гарантированное падение каждый месяц.

### Возможность реактивации
По аналитическому правилу 90 дней 961 клиент (16,3% клиентской базы) попал в сегмент High-value inactive. На этих клиентов приходится 15,5% исторической выручки. Это потенциальный сегмент для проверки гипотезы о реактивации, но 90 дней не является доказанным порогом churn.

### Концентрация выручки
Выручка заметно концентрирована среди наиболее ценных клиентов: top 10% клиентов формируют 63,9% выручки, а top 20% — 77,3%. Поэтому анализ среднего клиента может скрывать существенную разницу в экономической ценности клиентской базы.

## Проверка гипотез

- **H1 — поддерживается описательными данными:** 27,6% клиентов совершили только одну покупку.
- **H2 — частично поддерживается:** retention на длинном горизонте ниже начальных значений, но помесячная динамика не является монотонной.
- **H3 — поддерживается описательными данными:** repeat-клиенты составляют 72,4% базы, но формируют 96,8% выручки; top 10% клиентов формируют 63,9% выручки.

## Ограничения

- Retention-анализ является описательным и не доказывает причинно-следственные связи.
- Новые когорты имеют более короткое окно наблюдения и поэтому напрямую несопоставимы со зрелыми когортами на длинном горизонте.
- Порог 90 дней для inactivity — аналитическое правило, а не подтверждённое определение churn.
- Клиенты без Customer ID не могут быть включены в customer-level анализ retention.
- Отмены и строки с неположительным количеством или ценой исключены из метрик положительных продаж и retention, но сами отмены остаются доступными для отдельного операционного анализа.

## Возможные следующие шаги

- Определить churn на основе типичного цикла повторной покупки, а не фиксированного порога в 90 дней.
- Сравнить retention и клиентскую ценность по странам или категориям товаров.
- Проверить, какие товары или категории чаще предшествуют повторной покупке.
- После формирования надёжной целевой переменной построить модель вероятности повторной покупки.
- Проверить гипотезы по улучшению retention с помощью контролируемых экспериментов.
